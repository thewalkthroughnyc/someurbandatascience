# Day 1 — Python Fundamentals

**Goal:** Get comfortable with the basic building blocks of Python — the things every program uses, no matter how complex.

By the end of this notebook you will be able to:
- Store and label values with **variables**
- Know the four most common **types** (int, float, str, bool)
- Collect ordered data in a **list**
- Collect labeled data in a **dict** (dictionary)
- Repeat actions with a **for loop**
- Package reusable logic in a **function**

Every example is anchored to Bushwick real estate or the B38 bus so the numbers mean something.

```python
# CELL 2 — Variables and types
#
# A variable is just a labeled box. You put a value in, give it a name,
# and refer to it by that name later. Python figures out the type automatically.

address      = "441 Wilson Ave, Brooklyn, NY"   # str  — text, always in quotes
year_built   = 1931                              # int  — whole number
lot_area_sf  = 2500                              # int  — square feet
assessed_val = 847000.0                          # float — number with a decimal
is_landmarked = False                            # bool — True or False, capital T/F

# type() tells you what kind of thing a variable holds
print(type(address))       # <class 'str'>
print(type(year_built))    # <class 'int'>
print(type(assessed_val))  # <class 'float'>
print(type(is_landmarked)) # <class 'bool'>

# f-strings let you embed variables inside text (the f before the quote activates it)
print(f"{address} was built in {year_built}.")
```

```python
# CELL 3 — Cap rate calculation
#
# Cap rate (capitalization rate) = Net Operating Income / Purchase Price
# It's a yield ratio — like asking "what percentage return does this building pay?"
# A 5% cap means you'd earn back your investment in 20 years if income held flat.
#
# 441 Wilson Ave, Bushwick: a hypothetical 6-unit mixed-use walk-up.

purchase_price    = 1_850_000   # underscores are legal in numbers — easier to read
gross_rent_annual =   108_000   # 6 units × $1,500/mo × 12
operating_expenses =   37_800   # taxes, insurance, repairs (~35% of gross)

# Step-by-step (explicit)
noi = gross_rent_annual - operating_expenses
cap_rate_decimal = noi / purchase_price
cap_rate_pct     = cap_rate_decimal * 100

print(f"NOI:       ${noi:,.0f}")
print(f"Cap rate:  {cap_rate_pct:.2f}%")

# One-liner version — same math, fewer intermediate variables
# Useful once you're comfortable; harder to read when you're learning
print(f"Quick:     {(gross_rent_annual - operating_expenses) / purchase_price * 100:.2f}%")
```

```python
# CELL 4 — Lists
#
# A list is an ordered collection. Think of it like a numbered roster.
# Lists use square brackets [ ] and items are separated by commas.
# Indexing starts at 0, not 1 — the first item is always [0].

bushwick_addresses = [
    "441 Wilson Ave",
    "1083 Broadway",
    "315 Stanhope St",
    "56 Wyckoff Ave",
    "690 Knickerbocker Ave",
]

print(bushwick_addresses[0])   # first item
print(bushwick_addresses[-1])  # last item (-1 counts from the end)
print(len(bushwick_addresses)) # how many items

# Add a new address
bushwick_addresses.append("22 Linden St")
print(f"After append: {len(bushwick_addresses)} addresses")

# Slice: get a sub-list (start inclusive, end exclusive — like Python's range)
first_three = bushwick_addresses[0:3]
print(first_three)
```

```python
# CELL 5 — Dictionaries
#
# A dict stores key → value pairs. Think of it like a lookup table or a gradebook:
# student name (key) → grade (value). Keys must be unique.
# Dicts use curly braces { } and colons to separate keys from values.
#
# Here: B38 stop name → estimated daily southbound trips (from GTFS schedule).

b38_daily_trips = {
    "DeKalb Av / Broadway":        88,
    "DeKalb Av / Wyckoff Av":      86,
    "DeKalb Av / Wilson Av":       85,   # closest stop to 441 Wilson Ave
    "DeKalb Av / Knickerbocker Av": 84,
    "DeKalb Av / Bushwick Av":     83,
    "Myrtle Av / Broadway":        80,
}

# Look up a single stop
print(b38_daily_trips["DeKalb Av / Wilson Av"])

# .get() is safer — returns None instead of crashing if the key doesn't exist
print(b38_daily_trips.get("DeKalb Av / Cypress Av", "stop not found"))

# Explore the dict
print(list(b38_daily_trips.keys()))    # all stop names
print(list(b38_daily_trips.values()))  # all trip counts
print(len(b38_daily_trips))            # number of stops
```

```python
# CELL 6 — for loop + if/else (control flow)
#
# A for loop runs the same block of code once per item in a collection.
# if/else lets you branch: do one thing when a condition is True, another when False.
#
# Here: label each B38 stop as high-frequency or low-frequency.
# Threshold: >= 85 trips/day = high frequency (roughly every 10 min peak).

FREQ_THRESHOLD = 85  # all-caps = convention for a constant that won't change

# Verbose version — clearest to read when you're new
for stop, trips in b38_daily_trips.items():   # .items() gives (key, value) pairs
    if trips >= FREQ_THRESHOLD:
        label = "HIGH-FREQUENCY"
    else:
        label = "low-frequency"
    print(f"{label:17s}  {trips:3d} trips/day  {stop}")

print()

# Idiomatic version — same result, using a ternary expression
# Read as: value_if_true  if condition  else value_if_false
for stop, trips in b38_daily_trips.items():
    label = "HIGH-FREQUENCY" if trips >= FREQ_THRESHOLD else "low-frequency"
    print(f"{label:17s}  {trips:3d} trips/day  {stop}")
```

```python
# CELL 7 — Functions
#
# A function packages reusable logic under a name. You define it once with `def`,
# then call it as many times as you need. Think of it like a formula you write
# on the board once and students apply to different numbers.
#
# Parameters are the inputs (what goes into the formula).
# return sends the result back to whoever called the function.

def cap_rate(noi, price):
    """Return cap rate as a percentage. noi and price must be in the same currency."""
    if price == 0:
        return None          # guard against division by zero
    return (noi / price) * 100


# Property A: 441 Wilson Ave (from Cell 3)
rate_wilson = cap_rate(noi=70_200, price=1_850_000)
print(f"441 Wilson Ave:          {rate_wilson:.2f}% cap rate")

# Property B: hypothetical 4-unit on Wyckoff Ave
rate_wyckoff = cap_rate(noi=48_000, price=1_100_000)
print(f"Wyckoff Ave 4-unit:      {rate_wyckoff:.2f}% cap rate")

# Which is the better yield?
better = "Wilson" if rate_wilson > rate_wyckoff else "Wyckoff"
print(f"Better cap rate:         {better}")

# Demonstrate the zero-guard
print(f"Price = 0 edge case:     {cap_rate(50_000, 0)}")
```

## Day 1 Recap

| Concept | Syntax | Used for |
|---------|--------|----------|
| Variable | `x = 42` | Label a value |
| Types | `int`, `float`, `str`, `bool` | What kind of value |
| List | `[a, b, c]` | Ordered collection, indexed from 0 |
| Dict | `{key: value}` | Labeled lookup table |
| for loop | `for item in collection:` | Repeat for each element |
| if/else | `if cond: ... else: ...` | Branch on a condition |
| Function | `def name(params): return ...` | Reusable formula |
