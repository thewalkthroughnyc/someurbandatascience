## Git vs GitHub — not the same thing

| | What it is | Analogy |
|---|---|---|
| **Git** | Software on your computer that tracks changes | Like "Track Changes" in Google Docs, but for code |
| **GitHub** | A website that stores git history in the cloud | Like Google Drive, but for git repos |

You use git locally to save snapshots. You push to GitHub so the world (or future you, on a different laptop) can access them.

## The four git commands you'll use every day

```
git status          — what's changed since my last save?
git add <file>      — mark this file to be included in the next save
git commit -m "..." — save a snapshot with a note about what changed
git push            — send my snapshots to GitHub
```

## MANUAL STEP — Create a repo on GitHub

Git is local. To put it on GitHub, you need to create a repo there first.

1. Go to **github.com** — sign in or create a free account
2. Click the **+** button (top right) → **New repository**
3. Fill in:
   - **Repository name:** `urban-data-science` (or whatever you want)
   - **Description:** `Self-study: NYC housing, transit, and neighborhood change`
   - **Public** (so it works as a portfolio piece)
   - **Do NOT check** "Add a README file" — you already have CLAUDE.md
4. Click **Create repository**
5. GitHub shows you a page with setup commands. Copy the URL that ends in `.git` — it looks like:
   `https://github.com/YOUR-USERNAME/urban-data-science.git`
6. Paste that URL into the cell below where it says `YOUR-GITHUB-URL`

````python
# CELL 7 — Connect your local repo to GitHub.
#
# git remote add origin <url> tells git: "when I push, send it here."
# 'origin' is just the conventional nickname for the remote.
#
# EDIT: replace the URL below with your actual GitHub repo URL.

!git remote add origin https://github.com/thewalkthroughnyc/someurbandatascience.git

# Verify it registered:
!git remote -v
````

````python
# CELL 8 — Push to GitHub.
#
# git push sends your commits to GitHub.
# -u origin main means: "set 'origin main' as the default, so future pushes
# just need 'git push' with no extra arguments."
#
# GitHub will ask for your username and password.
# NOTE: GitHub no longer accepts your account password — you need a Personal Access Token.
# If it fails with an authentication error, see the cell below for how to create one.

!git push -u origin main
````

## If the push failed: Personal Access Token

GitHub stopped accepting passwords in 2021. You need a token instead.

1. Go to **github.com → Settings** (click your profile photo, top right)
2. Scroll down to **Developer settings** (bottom of left sidebar)
3. Click **Personal access tokens → Tokens (classic)**
4. Click **Generate new token (classic)**
5. Give it a name (e.g., "macbook"), set expiration to 90 days
6. Check the **repo** scope (that's all you need)
7. Click **Generate token** — copy it immediately (you only see it once)
8. When git asks for your password, paste the token instead of your real password

Then rerun Cell 8.

````python
# CELL 9 — Verify it's live.
#
# If the push worked, your repo is now public at:
# https://github.com/YOUR-USERNAME/urban-data-science
#
# Let's also confirm git knows where to push in the future.

!git remote -v
!git log --oneline
print('\nOpen https://github.com/thewalkthroughnyc/someurbandatascience.git in your browser.')
````
