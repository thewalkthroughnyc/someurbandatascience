# 02 — Compute Gap

Spatial-joins each Bushwick PLUTO lot to its Census block group (point-in-polygon), then computes `far_gap` per lot.

Output (written to `cache/`): `bw_joined.geojson`


```python
import geopandas as gpd
```

```python
gdf_pluto = gpd.read_file('cache/bw_pluto.geojson')
gdf_bg = gpd.read_file('cache/bg_kings.geojson')
```

```python
# Spatial join: does each lot's point fall within a block group polygon?
gdf_joined = gpd.sjoin(gdf_pluto, gdf_bg, predicate='within')

# Sanity check: boundary-line lots can match more than one polygon and
# duplicate a row, same idea as checking len() before/after a regular merge.
print(len(gdf_pluto), '->', len(gdf_joined))
```

```python
gdf_joined['far_gap'] = (gdf_joined['residfar'] - gdf_joined['builtfar']).clip(lower=0)

gdf_joined.to_file('cache/bw_joined.geojson', driver='GeoJSON')
print(gdf_joined[['BBL', 'GEOID', 'far_gap']].head())
```
